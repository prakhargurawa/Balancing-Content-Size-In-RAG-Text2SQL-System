```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Data
percentage_edit_distance = [52.99,51.32,50.76,51.32,47.57,47.29,47.29]
percentage_embedding_match = [67.87,68.43,68.85,68.29,64.26,63.56,64.12]
percentage_fuzzy_match = [45.20,44.65,43.25,42.56,40.19,39.36,39.08]

# Create a DataFrame for Seaborn
df = pd.DataFrame({
    'Index': [1, 2, 3, 4, 5, 6, 7],
    'Edit Distance Similarity Score': percentage_edit_distance,
    'Embedding Similarity Score': percentage_embedding_match,
    'Fuzzy Similarity Score': percentage_fuzzy_match
})

# Plot the line chart
plt.figure(figsize=(8, 6))  # Set figure size for better readability

# Plot each dataset with its corresponding color
sns.lineplot(data=df, x='Index', y='Edit Distance Similarity Score', marker='o', color='royalblue', linewidth=2, markersize=8, label='Edit Distance Similarity Score')
sns.lineplot(data=df, x='Index', y='Embedding Similarity Score', marker='o', color='#C71585', linewidth=2, markersize=8, label='Embedding Similarity Score')
sns.lineplot(data=df, x='Index', y='Fuzzy Similarity Score', marker='o', color='#2E8B57', linewidth=2, markersize=8, label='Fuzzy Similarity Score')

# Change background color of the plot area (inside the plot)
plt.gca().set_facecolor('#FFF8E1')  # Lighter and brighter cream color

# Enhance graph appearance
plt.title('SQL Query Similarity Score Pattern over Data', fontsize=14, fontweight='bold')
plt.xlabel('Spider-Data', fontsize=12)
plt.ylabel('Similarity Score', fontsize=12)
plt.ylim(0, 100)  # Set y-axis scale from 0 to 100
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)
plt.grid(True, linestyle='--', alpha=0.7)  # Add grid lines with light dashed lines for better readability

# Add legend
plt.legend(fontsize=10, loc='upper left')

# Show the plot
plt.tight_layout()

# Save the plot to a specific location
save_path = 'C:\Research-Paper\PAPER-WORK-2024\Outputs\SQL-Similarity-Score-Analysis\plot-sql-query-match-score.png'
plt.savefig(save_path, bbox_inches='tight', dpi=300)  # Saves the plot with high resolution

plt.show()
```
